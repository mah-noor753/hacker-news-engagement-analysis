# 🚀 Hacker News Data Analysis & Engagement Insights
**Author:** Mahnoor  
**Tools:** Python (Pandas, Seaborn, Matplotlib), Google BigQuery, SQL  
**Objective:** Analyzing public BigQuery datasets to uncover engagement patterns and trends on Hacker News stories.

```python
from google.cloud import bigquery
import pandas as pd

client = bigquery.Client()

query = """
    SELECT title, score, descendants AS comment_count
    FROM `bigquery-public-data.hacker_news.full`
    WHERE type = 'story' AND score IS NOT NULL AND descendants IS NOT NULL
    ORDER BY score DESC
    LIMIT 10
"""

df = client.query(query).result().to_dataframe()
print('Top 10 Stories With Scores and Comments:')
display(df)
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(10, 6))
sns.barplot(x='score', y='title', data=df, color='skyblue')
plt.title('Top 10 Hacker News Stories by Score', fontsize=14, fontweight='bold')
plt.xlabel('Score (Upvotes)', fontsize=12)
plt.ylabel('Story Title', fontsize=12)
plt.tight_layout()
plt.show()
```

### 💡 Key Findings & Insights
- **Viral Content Drivers:** Major tech milestones, breaking news, and industry-shaking events capture the highest user upvotes and community engagement on Hacker News.
- **Technical Execution:** Successfully queried millions of rows from Google BigQuery using SQL, processed the data via Pandas, and built clean visual analytics using Python.
